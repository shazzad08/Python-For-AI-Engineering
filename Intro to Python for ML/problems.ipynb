# M. Replace MinMax
## https://codeforces.com/group/MWSDmqGsZm/contest/219774/problem/M


```python

n = int(input())
arry = []

values= input().split()

for i in range(n):
     x = int(values[i])
     arry.append(x)

min1=arry[0]; idx1=0
max1=arry[0]; idx2=0

for i in range(n):
    if arry[i]<min1:
        min1=arry[i]
        idx1= i

    if arry[i] > max1:
        max1= arry[i]
        idx2= i

arry[idx1]= max1
arry[idx2] = min1

print(*arry)
```

# H. Good or Bad
## https://codeforces.com/group/MWSDmqGsZm/contest/219856/problem/H

```python
t = int(input())
list =[]

while t>0 :
    x= input()
    t-=1
    list.append(x)

for i in range(len(list)):
    if "010" in list[i] or "101" in list[i]:
        print("Good")
    else:
        print("Bad")
    
    
```
