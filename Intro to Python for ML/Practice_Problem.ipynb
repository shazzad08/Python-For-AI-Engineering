# Problem 01 : Smart Light

```python
x,y = map(float , input().split())

if x>=y:
    print("ON")
else: 
    print("OFF")
```

# Problem 02 : Sensor Calibrator — Min-Max Scaling

```python
x,min_v,  max_v = map(float , input().split())

norm = (x - min_v) / (max_v - min_v)

print(f"{norm:.2f}")
```

# Problem 03 : Mini Loss Tracker

```python
n= int(input())

t = float(input())
sum1= 0
for i in range(n):
    x= float(input())
    sum1+=x
avg= sum1/n

if avg<= t:
    print("PASS")
else:
    print("RETRY")
    
```

# problem 04 : Majority Vote Gate

```python
n= int(input())

y=0;no=0

for i in range(n):
    s = input()
    if s=="YES":
        y+=1
    else:
        no+=1
if y >= no:
    print("ACCEPT")
else:
    print("REJECT")
```

# Problem 05 : AI Mood Detector

```python
s = input().lower()
if "happy" in s or "joy" in s or "smile" in s:
    print("Happy Mood")
elif "sad" in s or "cry" in s or "angry" in s:
    print("Sad Mood")
else:
    print("Neutral Mood")
```

# Problem 06 : AI Brightness Analyzer

```python
arry = list(map(int, input().split(" ")))

sum1=0
for i in range(len(arry)):
    sum1+=arry[i]
avg1 = sum1/len(arry)

if avg1 < 85:
    print("Dark Image")
elif avg1 > 170 :
    print("Bright Image")
else:
    print("Normal Image")
    
```

# Problem 07 : AI Bias Detector

```python
char = input().upper().split(" ")
cnt_A=0; cnt_B=0

for i in range(len(char)):
    if char[i]=='A':
        cnt_A+=1
        
    else:
        cnt_B+=1


fA= (cnt_A/len(char))*100
fB= (cnt_B/len(char))*100

if fA >=70 or fB>=70:
    print("Biased Model")
else:
    print("Fair Model")
```

# problem 08: AI Keyword Counter

```python
s= input().lower().split()
keywords = ["ai", "data", "model", "learn", "train", "neural"]
count = 0

for word in keywords:
    if word in s:
        count+=1
if count>=2:
    print("AI Detected")
else:
    print("Not AI Related")
```
