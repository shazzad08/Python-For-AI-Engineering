E. Lowest Number

```python

n = int(input())
list=[]

values= input().split()

for i in range(n):
    x = int(values[i])
    list.append(x)

print(list)

min = list[0]
idx = 0

for i in range(1,n):
    if list[i] < min:
        min= list[i]
        idx=i
print(min, idx+1)
```
