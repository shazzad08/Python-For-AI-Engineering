# J. Count Letters
## https://codeforces.com/group/MWSDmqGsZm/contest/219856/problem/J

```python
s= input().lower()
freq= {}

for ch in s:
    if ch in freq:
        freq[ch]+=1
    else :
        freq[ch]= 1

for ch,freq in freq.items():
    print(f"{ch} : {freq}")
```

# No Idea!
## https://www.hackerrank.com/challenges/no-idea/problem

```python
n,m = input().split()
arry = list(map(int, input().split(" ")))
a = set(map(int, input().split(" ")))
b = set(map(int, input().split(" ")))

h=0

for num in arry:
    if num in a:
        h+=1
    elif num in b:
        h-=1
print(h)

```
