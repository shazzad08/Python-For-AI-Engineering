```python
class Phone:
    def __init__(self,model,battery=100):
        self.model= model
        self.battery= battery

class SmartPhone (Phone):
    def __init__(self,model,battery,os):
        super().__init__(model,battery)
        self.os= os

class GamingPhone(SmartPhone):
    def __init__(self,model,battery,os,cooling_system):
        super().__init__(model,battery,os)
        self.cooling_system= cooling_system
    def start_game(self,name):
        print(f"Playing {name} on {self.model}")
        
 
        
```

```python
phone = GamingPhone("ROG Phone 9", 100, "Android", "Liquid Cooling")

phone.start_game("PUBG")
```
