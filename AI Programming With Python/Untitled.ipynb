# Problem 1

```python
name,age= input().split()

def introduce(x,y):
    print(f"My name is {x} and I am {y} years old.")

introduce(name,age)
```

# Problem 2

```python
lst = list(map(int,input().split()))
sz= int(input())

def batch_generator(lst1, sz1):
    for i in range(0,len(lst),sz1):
        yield lst1[i: i+ sz1]
x= batch_generator(lst,sz)

for i in range(sz):
    print(next(x))
```

# Problem 4



```python
lst = list(map(int,input().split()))

def fun_power(x,y=2):
    return x**y


if(len(lst)==1):
    print(fun_power(lst[0]))
else:
    print(fun_power(lst[0],lst[1]))
```

# Problem 5 ( Conceptual Session ) 

```python
lst= list(map(int, input().split()))

def solution(n):
    return n%2==0

ans= filter(solution,lst)

result = map(lambda x:x**2,list(ans))

print(list(result))
```
